## Decisions to clean up our data¶

```python
# DECISION 1: We will REMOVE YEAR and KEEP MONTH because we believe shark attacks are seasonal.
# DECISION 2: We will REMOVE NAME, SEX and AGE because they have no relevance to the shark's actions.
# DECISION 3: We will REMOVE LOCATION AND KEEP COUNTRY and STATE, which are sufficient for location based attack hypothesis.
# DECISION 4: We will KEEP ACTIVITY to determine if certain activities correlate with shark attacks.
# DEICION 5: We will KEEP FATAL and SPECIES because these are both relevant for our hypothesis around species, attacks, and deaths.
# DECISION 5: We will REMOVE TIME and REMOVE ALL COLUMNS FROM SOURCE ONWARD because of missing or unnecessary data.

# FINAL DATA: Date, Type, Country, State, Activity, Fatal
### No problem with data types besides DATES!!!!
#Data Types = object => need to determin data within
# Country = string
# State = string
# Activity = string
# Fatal = object
# Species = string
```

```python
#REMOVE Column irrelevant to hypothesis
shark_col_drop = shark_data.drop(columns=["Year" ,"Location", "Name" ,"Sex" ,"Age" ,"Injury" , "Time" , "Source" , "pdf" ,"href formula" , "href" , "Case Number" , "Case Number.1" , "original order" , "Unnamed: 21", "Unnamed: 22"])
display(shark_col_drop.head())
```

```python
#Number of unique values per label
shark_unique = shark_col_drop.nunique()
display(shark_unique)
```

Cleaning of "Type" column because lower number of unique values

```python
#Review the unique values
shark_col_drop.columns = (
    shark_col_drop.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(shark_col_drop.columns.tolist())
```

```python
### CLEANING TO BE DONE , Part 2#

 #Type = correct mistakes & create "unknown" grouping, convert
 #Country = lower().replace(" ", "_").strip()
 #State = lower().replace(" ", "_").strip()
 #Activity = fuzzy connections ("swim") = ("swimming")
 #Fatal = clean to "yes", "no", "unknown"
 #Species = fuzzy connections ("swim") = ("swimming")
```

```python
shark_col_drop["type"].unique().tolist()
print(shark_col_drop["type"].unique().tolist())
shark_col_drop["fatal_y/n"].unique().tolist()
print(shark_col_drop["fatal_y/n"].unique().tolist())
```

```python
#COLUMN CLEANING
shark_col_drop["type"] = (
    shark_col_drop["type"]
    .str.lower()
    .str.strip()
)

display(shark_col_drop["type"].unique().tolist())
```

```python
#CONSTRUCT A DICTIONARY TO CLEAN 'TYPE', AND APPLY DICTIONARY TO 'TYPE' VALUES
shark_type_nan = shark_col_drop.dropna(subset = ["type"])
type_dictionary = {
    "unprovoked": "unprovoked",
    "provoked": "provoked",
    "questionable": "unknown_reason",
    "watercraft": "unknown_reason",
    "sea disaster": "unprovoked",
    "?" : "unknown_reason",
    "unconfirmed": "unknown_reason",
    "unverified": "unknown_reason",
    "invalid": "unknown_reason",
    "under investigation": "unknown_reason",
    "boat": "unknown_reason"
}

shark_type_nan["type_clean"] = shark_type_nan["type"].map(type_dictionary)
#display(shark_type_nan["type_clean"].unique().tolist())
display(shark_type_nan.head())
shark_type_clean = (
    shark_type_nan.drop(columns=["type"])
    .rename(columns={"type_clean": "type"})
    .iloc[ : , [ 0 , -1, 1 , 2 ,3 , 4, 5] ]
)
display(shark_type_clean)
```

```python
# Cleaning 'country' and 'state' and 'fatal' data:
shark_type_clean["country"] = (
    shark_type_clean["country"]
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

shark_type_clean["state"] = (
    shark_type_clean["state"]
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

shark_type_clean["fatal_y/n"] = (
    shark_type_clean["fatal_y/n"]
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(shark_type_clean["fatal_y/n"].unique().tolist())
display(shark_type_clean["fatal_y/n"].value_counts(dropna=False))

shark_type_clean["fatal_y/n"] = shark_type_clean["fatal_y/n"].fillna("unknown")
display(shark_type_clean["fatal_y/n"].unique().tolist())

fatal_dictionary = {
    "n": "no",
    "y": "yes",
    "m": "unknown",
    "unknown": "unknown",
    "f": "unknown",
    "nq" : "unknown",
    "y_x_2": "unknown",
}

shark_type_clean["fatal_y/n_clean"] = shark_type_clean["fatal_y/n"].map(fatal_dictionary)
display(shark_type_clean["fatal_y/n_clean"].unique().tolist())

#Remove original fatal_y/n and replace and rename fatal_y/n_clean
shark_csf_clean = (
    shark_type_clean.drop(columns=["fatal_y/n"])
    .rename(columns={"fatal_y/n_clean": "fatal_y/n"})
    .iloc[ : ,[ 0 , 1, 2 , 3 , 4 , -1, 5] ]
)
display(shark_csf_clean)
shark_data_insights = shark_csf_clean
```

```python
display(shark_data_insights)
```

```python
# Cleaning the "activity" column

shark_data_insights["activity"].isna().sum()
shark_data_insights["activity"] = shark_data_insights["activity"].fillna("Other")
```

```python
def clean_activity(value):
    if "swim" in value.lower():
        return "Swimming"
    if "snork" in value.lower():
        return "Snorkeling"
    elif "fish" in value.lower():
        return "Fishing"
    elif "surf" in value.lower():
        return "Surfing"
    elif "div" in value.lower():
        return "Diving"
    elif "wad" in value.lower():
        return "Wading"
    elif "stand" in value.lower():
        return "Standing"
    elif "bath" in value.lower():
        return "Bathing"
    elif "over" in value.lower():
        "Overboard"
    elif "boat" in value.lower():
        "Boating"
    elif "disaster" in value.lower():
        "Accident"
    elif "accident" in value.lower():
        "Accident"
    else:
        return "Other"
```

```python
shark_data_insights["cleaned_activity"] = shark_data_insights["activity"].apply(clean_activity)
shark_data_insights["cleaned_activity"].value_counts()
```

```python
display(shark_data_insights.head())
```

```python
# Removing uncleaned data columns :
shark_data_final = shark_data_insights.drop(columns=["date", "species", "activity"])
display(shark_data_final.head())
```

```python
shark_data_final = shark_data_final.rename(columns=
                                           {"cleaned_activity": "activity",
                                           "fatal_y/n" : "fatal"}
                                          )
display(shark_data_final.head())
```

```python
#Final look of the data frame :
shark_data_final = shark_data_final.iloc[ : , [1, 2, -1, 0, -2]]
display(shark_data_final.head())
```

```python
country_state_counts = (
    shark_data_final.groupby(["country", "state"])
    .size()
    .sort_values(ascending=False)
)

display(country_state_counts.head(20))
```

```python
country_state_counts = (
    shark_data_final.groupby(["country", "state"])
    .size()
)

display(country_state_counts)
```

```python
country_totals = shark_data_final["country"].value_counts()

country_state_counts = (
    shark_data_final.groupby(["country", "state"])
    .size()
    .reset_index(name="state_attacks")
)

country_state_counts["country_total"] = (
    country_state_counts["country"].map(country_totals)
)

country_state_counts = country_state_counts.sort_values(
    ["country_total", "state_attacks"],
    ascending=[False, False]
)

top_10 = country_totals.head(10).index

result = country_state_counts[
    country_state_counts["country"].isin(top_10)
]

print(result.to_string(index=False))
```

```python
# Find the 10 countries with the most shark attacks.

top_countries = shark_data_final["country"].value_counts().head(10)

# Find the top 5 states for each country.

tables = []

for country in top_countries.index:

    top_states = (
        shark_data_final[shark_data_final["country"] == country]["state"]
        .value_counts()
        .head(5)
        .reset_index()
    )

    top_states.columns = ["State", "State attacks"]

    top_states["Country"] = country
    top_states["Country attacks"] = top_countries[country]

    tables.append(top_states)

# Combine everything into ONE table.

result = pd.concat(tables, ignore_index=True)

result = result[["Country", "Country attacks", "State", "State attacks"]]

# Display each country and its total only once.

display(result.set_index(["Country", "Country attacks", "State"]))
```

```python
#  Select unprovoked attacks from our existing dataset.

unprovoked = shark_data_final[
    shark_data_final["type"] == "unprovoked"
]


# Count unprovoked attacks for each country.

country_unprovoked = unprovoked["country"].value_counts()

result["Country unprovoked"] = (
    result["Country"].map(country_unprovoked).fillna(0).astype(int)
)


# Count unprovoked attacks for each country and state.

state_unprovoked = (
    unprovoked.groupby(["country", "state"])
    .size()
    .reset_index(name="State unprovoked")
)


# Add those counts to our EXISTING table.

result = result.merge(
    state_unprovoked,
    left_on=["Country", "State"],
    right_on=["country", "state"],
    how="left"
)

result = result.drop(columns=["country", "state"])

result["State unprovoked"] = (
    result["State unprovoked"].fillna(0).astype(int)
)


# Display the same table with unprovoked information.

display(result.set_index([
    "Country",
    "Country attacks",
    "Country unprovoked",
    "State"
]))
```
