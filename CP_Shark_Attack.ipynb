```python
#COLUMN CLEANING
shark_col_drop.columns = (
    shark_col_drop.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(shark_col_drop.columns.tolist())
```

```python
#DEEP DIVE UNIQUE VALUES IN TYPE & FATAL Y/N - SMALLEST NUMBER OF UNIQUE VALUES

print(f"The unique values of 'type' are: ")
display(shark_col_drop["type"].unique().tolist())

print(f"\nThe unique values of Fatal attacks are: ")
display(shark_col_drop["fatal_y/n"].unique().tolist())
```

```python
# Cleaning 'type' data
shark_col_drop["type"] = (
    shark_col_drop["type"]
    .str.lower()
    .str.strip()
)

display(shark_col_drop["type"].unique().tolist())
```

```python
# REMOVE NaN FROM TYPE 
shark_type_nan = shark_col_drop.dropna(subset = ["type"])

display(shark_type_nan["type"].unique().tolist())
```

```python
#CONSTRUCT A DICTIONARY TO CLEAN 'TYPE', AND APPLY DICTIONARY TO 'TYPE' VALUES

print(type(shark_type_nan["type"]))

type_dictionary = {
    "unprovoked": "unprovoked",
    "provoked": "provoked",
    "questionable": "unknown_reason",
    "watercraft": "unknown_reason",
    "sea disaster": "unprovoked",
    "?" : "unknown_reason",
    "unconfirmed": "unknown_reason",
    "unverified": "unknown_reason",
    "invalid": "unknown_reason",
    "under investigation": "unknown_reason",
    "boat": "unknown_reason"
}

shark_type_nan["type_clean"] = shark_type_nan["type"].map(type_dictionary)
#display(shark_type_nan["type_clean"].unique().tolist())
display(shark_type_nan.head())
shark_type_clean = (
    shark_type_nan.drop(columns=["type"])
    .rename(columns={"type_clean": "type"})
    .iloc[ : ,  [ 0 , -1, 1 , 2 ,3 , 4, 5] ]
                 )
display(shark_type_clean)
```

```python
### COMPLETED CLEANING
#1. Type = correct mistakes & create "unknown" grouping, convert

### CLEANING TO BE DONE
# Date = Regex and Date Functions
# Country = lower().replace(" ", "_").strip()
# State = lower().replace(" ", "_").strip()
# Activity = fuzzy connections ("swim") = ("swimming")
# Fatal = clean to "yes", "no", "unknown"
# Species = fuzzy connections ("swim") = ("swimming")
```

```python
# Cleaning 'country' and 'state' and 'fatal' data:
shark_type_clean["country"] = (
    shark_type_clean["country"]
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# Cleaning 'state' data
shark_type_clean["state"] = (
    shark_type_clean["state"]
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

#3 Cleaning 'fatal' data:
shark_type_clean["fatal_y/n"] = (
    shark_type_clean["fatal_y/n"]
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(shark_type_clean["fatal_y/n"].unique().tolist())
display(shark_type_clean["fatal_y/n"].value_counts(dropna=False))
```

```python
#Fatal data cleaning decisions
# y = yes
# n = no
# NaN = unknown
# unknown = unknown
# f = unknown
# nq = unknown
# y_x_2 = unknown

#Replace missing values with "unknown"

shark_type_clean["fatal_y/n"] = shark_type_clean["fatal_y/n"].fillna("unknown")
display(shark_type_clean["fatal_y/n"].unique().tolist())

fatal_dictionary = {
    "n": "no",
    "y": "yes",
    "m": "unknown",
    "unknown": "unknown",
   "f": "unknown",
    "nq" : "unknown",
    "y_x_2": "unknown",
}

shark_type_clean["fatal_y/n_clean"] = shark_type_clean["fatal_y/n"].map(fatal_dictionary)
display(shark_type_clean["fatal_y/n_clean"].unique().tolist())

#Check for success
#display(shark_type_clean.head(10))

#Remove original fatal_y/n and replace and rename fatal_y/n_clean
shark_csf_clean = (
    shark_type_clean.drop(columns=["fatal_y/n"])
    .rename(columns={"fatal_y/n_clean": "fatal_y/n"})
    .iloc[ : ,  [ 0 , 1, 2 , 3 , 4 , -1, 5] ]
                 )
display(shark_csf_clean)
```
